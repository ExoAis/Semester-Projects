```python
import numpy as np
```

```python
num_states = 3
num_actions = 2
discount_factor = 0.9
```

```python
transitions = np.zeros((num_states, num_actions, num_states))
rewards = np.zeros((num_states, num_actions, num_states))
q_values = np.zeros((num_states, num_actions))
transitions[0, 0, 0] = 0.5 # Top-Drive-Top
transitions[0, 0, 1] = 0.5 # Top-Drive-Rolling
transitions[0, 1, 0] = 0.5 # Top-No Drive-Top
transitions[0, 1, 1] = 0.5 # Top-No Drive-Rolling
transitions[1, 0, 0] = 0.3 # Rolling-Drive-Top
transitions[1, 0, 1] = 0.4 # Rolling-Drive-Rolling
transitions[1, 0, 2] = 0.3 # Rolling-Drive-Bottom
transitions[1, 1, 2] = 1 # Rolling-No Drive-Bottom
transitions[2, 0, 0] = 0.5 # Bottom-Drive-Top
transitions[2, 0, 2] = 0.5 # Bottom-Drive-Bottom
transitions[2, 1, 2] = 1 # Bottom-No Drive-Bottom
rewards[0, 0, 0] = 2 # Top-Drive-Top
rewards[0, 0, 1] = 2 # Top-Drive-Rolling
rewards[0, 1, 0] = 3 # Top-No Drive-Top
rewards[0, 1, 1] = 1 # Top-No Drive-Rolling
rewards[1, 0, 0] = 2 # Rolling-Drive-Top
rewards[1, 0, 1] = 1.5 # Rolling-Drive-Rolling
rewards[1, 0, 2] = 0.5 # Rolling-Drive-Bottom
rewards[1, 1, 2] = 1 # Rolling-No Drive-Bottom
rewards[2, 0, 0] = 2 # Bottom-Drive-Top
rewards[2, 0, 2] = 2 # Bottom-Drive-Bottom
rewards[2, 1, 2] = 1 # Bottom-No Drive-Bottom
```

```python
def policy_iteration_stochastic(q_values, transitions, rewards, discount_factor):
    num_states, num_actions, _ = transitions.shape
    values = np.zeros(num_states)
    policy = np.zeros(num_states, dtype = int)
    initial = True
    while True:
        delta = 0
        for s in range(num_states):
            v = values[s]
            if initial:
                value_1 = 0.5 * sum(transitions[s, 0, s_next] * (rewards[s, 0, s_next] + discount_factor * values[s_next]) for s_next in range(num_states))
                value_2 = 0.5 * sum(transitions[s, 1, s_next] * (rewards[s, 1, s_next] + discount_factor * values[s_next]) for s_next in range(num_states))
                values[s] = value_1 + value_2
                initial = False
            else:
                action = policy[s]
                values[s] = sum(transitions[s, action, s_next] * (rewards[s, action, s_next] + discount_factor * values[s_next]) for s_next in range(num_states))
        policy_stable = True
        for s in range(num_states):
            old_action = policy[s]
            for a in range(num_actions):
                q_values[s, a] = sum(transitions[s, a, s_next] * (rewards[s, a, s_next] + discount_factor * values[s_next]) for s_next in range(num_states))
            policy[s] = np.argmax(q_values[s])
            if old_action != policy[s]:
                policy_stable = False
        if policy_stable:
            break
    return values, q_values
```

```python
def optimal_policy(q_values):
    states = ['Top', 'Rolling', 'Bottom']
    actions = ['Drive', 'No Drive']
    policy = {}
    for s in range(q_values.shape[0]):
        policy[states[s]] = actions[np.argmax(q_values[s, :])]
    for state in policy.keys():
        print('From', state, ',', policy[state])
```

```python
values, q_values = policy_iteration_stochastic(q_values, transitions, rewards, discount_factor)
print('The values for each state after the value iteration are', values)
```

```python
optimal_policy(q_values)
```
