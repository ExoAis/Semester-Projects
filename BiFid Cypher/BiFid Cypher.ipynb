```python
import math

def encrypt(plain_text, key):
    rows = math.ceil(len(plain_text) / len(key))
    cols = len(key)
    grid = [['_' for i in range(cols)] for j in range(rows)]
    length = 0
    for i in range(rows):
        for j in range(cols):
            if length == len(plain_text):
                break
            grid[i][j] = plain_text[length]
            length += 1
    key_split = [i for i in key]
    sorted_keys = key_split.copy()
    sorted_keys.sort()
    positions = {}
    for k in key_split:
        positions[k] = sorted_keys.index(k)
    cipher = ''
    order = list(positions.values())
    for j in range(len(order)):
        idx = order.index(j)
        for i in range(rows):
            cipher += grid[i][idx]
    return cipher

def decrypt(cipher, key):
    rows = math.ceil(len(cipher) / len(key))
    cols = len(key)
    grid = [['_' for i in range(cols)] for j in range(rows)]
    length = 0
    for i in range(cols):
        for j in range(rows):
            if length == len(cipher):
                break
            grid[j][i] = cipher[length]
            length += 1
    key_split = [i for i in key]
    sorted_keys = key_split.copy()
    sorted_keys.sort()
    positions = {}
    for k in key_split:
        positions[k] = sorted_keys.index(k)
    order = positions.values()
    plain_text = ''
    for i in range(rows):
        for j in order:
            if grid[i][j] == '_':
                continue
            plain_text += grid[i][j]
    return plain_text

key = 'GREAT'
cipher = encrypt('Believe you can and you are halfway there', key)
print(cipher)
plain_text = decrypt(cipher, key)
print(plain_text)
```
