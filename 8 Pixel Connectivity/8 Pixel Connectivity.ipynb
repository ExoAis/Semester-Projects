```python
import numpy as np
import cv2
```

```python
image = cv2.imread('Connectivity.bmp')
image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) 
```

```python
def find_root(label):
    while label >= 0 and equivalency_table[label] >= 0:
        label = equivalency_table[label]
    return label
```

```python
labelled_image = np.zeros(image.shape, dtype = int)
equivalency_table = [-1]
h = image.shape[0]
w = image.shape[1]
colors = [(255, 0, 0), (0, 255, 0), (0, 0, 255), (255, 165, 0), (255, 192, 203), (255, 255, 0), (255, 255, 255), (128, 0, 128), (255, 215, 0), (153, 255, 255), (160, 160, 160)]
label = -1
#Given image would be in the range of 0-255 where to-be-segmented portions are of value 255 and the rest are of value 0
for i in range(h):
    for j in range(w):
        if image[i][j] == 255:
            neighbors = [labelled_image[i][j - 1], labelled_image[i - 1][j], labelled_image[i - 1][j - 1], labelled_image[i - 1][j + 1]]
            non_zero_neighbors = [i for i in neighbors if i != 0]
            if not non_zero_neighbors:
                label += 1
                labelled_image[i][j] = label
                equivalency_table.append(-1)
            else:
                min_label = min(non_zero_neighbors)
                labelled_image[i][j] = min_label
                for neighbor in non_zero_neighbors:
                    if neighbor != min_label:
                        equivalency_table[neighbor] = min_label
                        
for i in range(labelled_image.shape[0]):
    for j in range(labelled_image.shape[1]):
        if labelled_image[i][j] != 0:
            labelled_image[i][j] = find_root(labelled_image[i][j])

color_image = np.zeros((h, w, 3), dtype=np.uint8)
for i in range(h):
    for j in range(w):
        if labelled_image[i, j] != 0:
            label = labelled_image[i, j]
            color = colors[label % len(colors)]
            color_image[i, j] = color

cv2.imshow('Connected Components', color_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
